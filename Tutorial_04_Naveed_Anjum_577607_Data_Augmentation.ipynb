## **Tutorial 4: Data Augmentation**

```python
import os
import torch
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt

# Step 2: Define the Folder Where Augmented Images Will Be Saved
save_folder = r"/content/Augmented Images"
os.makedirs(save_folder, exist_ok=True)

# Step 3: Initialize the PyTorch Transforms Pipeline with Augmentation Parameters
# - rotation_range=40 -> RandomRotation(40)
# - shear_range=0.2 & zoom_range=0.2 -> RandomAffine (shear=20 degrees, scale=(0.8, 1.2))
# - horizontal_flip=True -> RandomHorizontalFlip (p=0.5 probability)
# - brightness_range=(0.5, 1.5) -> ColorJitter(brightness=(0.5, 1.5))
augmentation_transforms = transforms.Compose([
    transforms.RandomRotation(degrees=40),
    transforms.RandomAffine(degrees=0, shear=20, scale=(0.8, 1.2)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(brightness=(0.5, 1.5))
])

# Step 4: Load the Sample Image and Convert to PIL Image Format (Force RGB Mode to avoid mode errors)
image_path = r"/content/Image.jpg"
img = Image.open(image_path).convert("RGB")

# Steps 5 & 6: Generate, Save, and Display Augmented Images (Looping 20 times)
for i in range(1, 21):
    # Apply the augmentation pipeline to the image
    augmented_image = augmentation_transforms(img)

    # Save the augmented image to the specified directory
    save_path = os.path.join(save_folder, f"image_{i}.jpeg")
    augmented_image.save(save_path, "JPEG")

    # Display the augmented image using matplotlib (optional)
    plt.figure()
    plt.imshow(augmented_image)
    plt.axis('off')  # Hide axes for better visualization
    plt.show()


# Step 7: Confirm the Saved Location of Augmented Images
print(f"Augmented images are saved in the folder: {os.path.abspath(save_folder)}")

```
